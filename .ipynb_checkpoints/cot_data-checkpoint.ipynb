```python
import cot_reports as cot_r
import yfinance as yf
import pandas as pd
import numpy as np

full_dfs = []
for year in range(2000, 2026):
    try:
        df_year = cot_r.cot_year(year=year, cot_report_type='legacy_fut')
        full_dfs.append(df_year)
        print(f'got {year}: {len(df_year)} rows')
    except Exception as e:
        print(f'failed {year}: {e}')

full = pd.concat(full_dfs, ignore_index=True)

# Parse dates properly
full['date'] = pd.to_datetime(full['As of Date in Form YYMMDD'].astype(str).str.zfill(6), format='%y%m%d')

print(f'\nTotal rows: {len(full)}')
print(f'Date range: {full["date"].min()} to {full["date"].max()}')
```

```python
def cot_index(series, window=52):
    rolling_min = series.rolling(window).min()
    rolling_max = series.rolling(window).max()
    return (series - rolling_min) / (rolling_max - rolling_min)
```

```python
cot_markets = [
    'CORN - CHICAGO BOARD OF TRADE',
    'SOYBEANS - CHICAGO BOARD OF TRADE',
    'LEAN HOGS - CHICAGO MERCANTILE EXCHANGE',
    'CRUDE OIL, LIGHT SWEET - NEW YORK MERCANTILE EXCHANGE',
    "CRUDE OIL, LIGHT 'SWEET' - NEW YORK MERCANTILE EXCHANGE",
    'WTI FINANCIAL CRUDE OIL - NEW YORK MERCANTILE EXCHANGE',
]
cols = [
    'Market and Exchange Names',
    'date',
    'Commercial Positions-Long (All)',
    'Commercial Positions-Short (All)',
    'Noncommercial Positions-Long (All)',
    'Noncommercial Positions-Short (All)',
    'Open Interest (All)',
    'Change in Commercial-Long (All)',
    'Change in Commercial-Short (All)',
    'Change in Noncommercial-Long (All)',
    'Change in Noncommercial-Short (All)',
]

full['Market and Exchange Names'] = full['Market and Exchange Names'].str.strip()


cot = full[full['Market and Exchange Names'].isin(cot_markets)][cols].copy()
cot['Market and Exchange Names'] = cot['Market and Exchange Names'].replace({
    "CRUDE OIL, LIGHT 'SWEET' - NEW YORK MERCANTILE EXCHANGE": 'CRUDE OIL, LIGHT SWEET - NEW YORK MERCANTILE EXCHANGE',
    'WTI FINANCIAL CRUDE OIL - NEW YORK MERCANTILE EXCHANGE': 'CRUDE OIL, LIGHT SWEET - NEW YORK MERCANTILE EXCHANGE',
})

cot['tradeable_date'] = cot['date'] + pd.Timedelta(days=9)
cot['tradeable_date'] = cot['tradeable_date'].astype('datetime64[s]')
cot['date'] = pd.to_datetime(cot['date'])
cot = cot.sort_values(['Market and Exchange Names', 'date'])

cot['comm_net'] = cot['Commercial Positions-Long (All)'] - cot['Commercial Positions-Short (All)']
cot['spec_net'] = cot['Noncommercial Positions-Long (All)'] - cot['Noncommercial Positions-Short (All)']
for col in ['Change in Commercial-Long (All)', 'Change in Commercial-Short (All)',
            'Change in Noncommercial-Long (All)', 'Change in Noncommercial-Short (All)']:
    cot[col] = pd.to_numeric(cot[col], errors='coerce')
cot['comm_change'] = cot['Change in Commercial-Long (All)'] - cot['Change in Commercial-Short (All)']

cot['cot_idx'] = cot.groupby('Market and Exchange Names')['comm_net'].transform(
    lambda x: cot_index(x, window=52)
)
cot['spec_idx'] = cot.groupby('Market and Exchange Names')['spec_net'].transform(
    lambda x: cot_index(x, window=52)
)
cot['oi_idx'] = cot.groupby('Market and Exchange Names')['Open Interest (All)'].transform(
    lambda x: cot_index(x, window=52)
)

cot['tradeable_date'] = (cot['date'] + pd.Timedelta(days=9)).astype('datetime64[s]')

print(cot.groupby('Market and Exchange Names')['date'].agg(['min', 'max', 'count']))
print(cot[['cot_idx', 'spec_idx', 'oi_idx']].describe())
```

```python
cot.to_csv('cot_clean.csv', index=False)
```
