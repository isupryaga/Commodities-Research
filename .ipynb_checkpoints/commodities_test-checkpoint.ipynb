```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt

# Load COT
cot = pd.read_csv('cot_clean.csv', parse_dates=['date', 'tradeable_date'])

# Load EOD price data
tickers = {'corn': 'ZS=F', 'soy': 'ZC=F', 'hogs': 'HE=F', 'crude': 'CL=F'}
eod = {}
for name, ticker in tickers.items():
    df = yf.download(ticker, start='2000-01-01', auto_adjust=True)
    df.columns = ['_'.join(col).strip('_') for col in df.columns]
    eod[name] = df
    print(f"{name}: {len(df)} rows")

# Load intraday hourly
intraday = {}
for name, ticker in tickers.items():
    df = yf.download(ticker, interval='1h', period='2y', auto_adjust=True)
    df.columns = ['_'.join(col).strip('_') for col in df.columns]
    intraday[name] = df
    print(f"{name} 1h: {len(df)} rows")
```

```python
cot_market_map = {
    'CORN - CHICAGO BOARD OF TRADE': ('corn', 'ZS=F'),
    'SOYBEANS - CHICAGO BOARD OF TRADE': ('soy', 'ZC=F'),
    'LEAN HOGS - CHICAGO MERCANTILE EXCHANGE': ('hogs', 'HE=F'),
    'CRUDE OIL, LIGHT SWEET - NEW YORK MERCANTILE EXCHANGE': ('crude', 'CL=F'),
}

merged_1h = {}
for cot_name, (short_name, ticker) in cot_market_map.items():
    cot_market = cot[cot['Market and Exchange Names'] == cot_name][
        ['tradeable_date', 'cot_idx', 'spec_idx', 'oi_idx']
    ].copy()
    cot_market['tradeable_date'] = cot_market['tradeable_date'].astype('datetime64[s]')
    
    price = intraday[short_name].copy().reset_index()
    price['Datetime'] = price['Datetime'].dt.tz_localize(None).astype('datetime64[s]')
    
    merged_1h[short_name] = pd.merge_asof(
        price.sort_values('Datetime'),
        cot_market.sort_values('tradeable_date'),
        left_on='Datetime',
        right_on='tradeable_date',
        direction='backward'
    ).set_index('Datetime')

for name, df in merged_1h.items():
    print(f"{name}: {len(df)} rows, cot_idx NaN: {df['cot_idx'].isna().sum()}")
```

```python
for name in ['corn', 'soy', 'hogs', 'crude']:
    ticker = {'corn': 'ZS=F', 'soy': 'ZC=F', 'hogs': 'HE=F', 'crude': 'CL=F'}[name]
    d = merged_1h[name].copy()
    
    d['ret'] = d[f'Close_{ticker}'].pct_change()
    d['date'] = d.index.date
    d['hour'] = d.index.hour
    d['signal'] = (d['cot_idx'] > 0.8).astype(int)
    
    # For each day, compute: open-to-close return, and max adverse excursion
    daily = d.groupby('date').agg(
        signal=('signal', 'first'),
        open=(f'Open_{ticker}', 'first'),
        close=(f'Close_{ticker}', 'last'),
        high=(f'High_{ticker}', 'max'),
        low=(f'Low_{ticker}', 'min'),
    )
    daily['oc_ret'] = (daily['close'] - daily['open']) / daily['open']
    daily['range'] = (daily['high'] - daily['low']) / daily['open']
    daily['up_move'] = (daily['high'] - daily['open']) / daily['open']
    daily['down_move'] = (daily['open'] - daily['low']) / daily['open']
    
    on = daily[daily['signal'] == 1]
    off = daily[daily['signal'] == 0]
    
    print(f"\n{'='*40}")
    print(f"{name.upper()}")
    print(f"{'Metric':<20} {'Signal':>10} {'No Signal':>10}")
    print(f"{'OC return':<20} {on['oc_ret'].mean():>10.5f} {off['oc_ret'].mean():>10.5f}")
    print(f"{'Up move from open':<20} {on['up_move'].mean():>10.5f} {off['up_move'].mean():>10.5f}")
    print(f"{'Down move from open':<20} {on['down_move'].mean():>10.5f} {off['down_move'].mean():>10.5f}")
    print(f"{'Daily range':<20} {on['range'].mean():>10.5f} {off['range'].mean():>10.5f}")
    print(f"{'Win rate':<20} {(on['oc_ret']>0).mean():>10.2%} {(off['oc_ret']>0).mean():>10.2%}")
```

```python
for name in ['corn', 'soy', 'hogs', 'crude']:
    ticker = {'corn': 'ZS=F', 'soy': 'ZC=F', 'hogs': 'HE=F', 'crude': 'CL=F'}[name]
    d = merged_1h[name].copy()
    d.index = pd.to_datetime(d.index)
    
    d['date'] = d.index.date
    d['hour'] = d.index.hour
    d['signal'] = (d['cot_idx'] > 0.8).astype(int)
    
    daily_open = d[d['hour'] == 14].groupby('date')[f'Open_{ticker}'].first()
    daily_close = d.groupby('date')[f'Close_{ticker}'].last()
    daily_signal = d.groupby('date')['signal'].first()
    first3_low = d[d['hour'].isin([14, 15, 16])].groupby('date')[f'Low_{ticker}'].min()
    first3_high = d[d['hour'].isin([14, 15, 16])].groupby('date')[f'High_{ticker}'].max()
    
    daily = pd.DataFrame({
        'open': daily_open,
        'close': daily_close,
        'signal': daily_signal,
        'first3_low': first3_low,
        'first3_high': first3_high,
    }).dropna()
    
    daily['pullback'] = (daily['open'] - daily['first3_low']) / daily['open']
    daily['breakout'] = (daily['first3_high'] - daily['open']) / daily['open']
    daily['oc_ret'] = (daily['close'] - daily['open']) / daily['open']
    
    on = daily[daily['signal'] == 1]
    off = daily[daily['signal'] == 0]
    
    print(f"\n{'='*40}")
    print(f"{name.upper()}")
    print(f"{'Metric':<25} {'Signal':>10} {'No Signal':>10}")
    print(f"{'Avg pullback from open':<25} {on['pullback'].mean():>10.5f} {off['pullback'].mean():>10.5f}")
    print(f"{'Avg breakout from open':<25} {on['breakout'].mean():>10.5f} {off['breakout'].mean():>10.5f}")
    print(f"{'Days price pulled back':<25} {(on['pullback']>0.001).mean():>10.2%} {(off['pullback']>0.001).mean():>10.2%}")
    print(f"{'OC return':<25} {on['oc_ret'].mean():>10.5f} {off['oc_ret'].mean():>10.5f}")
```

```python
def backtest_cot(df, long_thresh=0.8, short_thresh=0.2, silent=False):
    df = df.copy().dropna(subset=['cot_idx', 'ret'])
    df['signal'] = 0
    df.loc[df['cot_idx'] > long_thresh, 'signal'] = 1
    df.loc[df['cot_idx'] < short_thresh, 'signal'] = -1
    df['signal'] = df['signal'].shift(1)
    df['strat_ret'] = df['signal'] * df['ret']
    df['equity'] = (1 + df['strat_ret']).cumprod()
    
    total_ret = df['equity'].iloc[-1] - 1
    annual_ret = (1 + total_ret) ** (252 / len(df)) - 1
    sharpe = df['strat_ret'].mean() / df['strat_ret'].std() * (252 ** 0.5)
    max_dd = (df['equity'] / df['equity'].cummax() - 1).min()
    win_rate = (df[df['signal'] != 0]['strat_ret'] > 0).mean()
    n_trades = (df['signal'] != 0).sum()
    
    metrics = dict(annual=annual_ret, sharpe=sharpe, max_dd=max_dd, 
                   win_rate=win_rate, n_days=n_trades)
    
    if not silent:
        for k, v in metrics.items():
            print(f"{k}: {v:.3f}")
    
    return df, metrics

for name, df in merged.items():
    print(f"\n{'='*40}\n{name.upper()}")
    print(f"{'Period':<12} {'Annual':>8} {'Sharpe':>8} {'MaxDD':>8} {'WinRate':>8}")
    for period_name, (start, end) in periods.items():
        mask = (df.index >= start) & (df.index <= end)
        period_df = df[mask]
        if len(period_df) < 50:
            continue
        _, m = backtest_cot(period_df, silent=True)
        print(f"{period_name:<12} {m['annual']:>8.2%} {m['sharpe']:>8.3f} {m['max_dd']:>8.2%} {m['win_rate']:>8.2%}")
```

```python
def backtest_cot_sized(df, long_thresh=0.8, short_thresh=0.2, long_only=False,
                        target_vol=0.01, vol_window=20, silent=False):
    df = df.copy().dropna(subset=['cot_idx', 'ret'])
    
    df['signal'] = 0
    df.loc[df['cot_idx'] > long_thresh, 'signal'] = 1
    if not long_only:
        df.loc[df['cot_idx'] < short_thresh, 'signal'] = -1
    df['signal'] = df['signal'].shift(1)
    
    # Scale position size by realized vol
    df['vol'] = df['ret'].rolling(vol_window).std()
    df['size'] = (target_vol / df['vol']).clip(0.1, 3.0)
    df['size'] = df['size'].shift(1)  # no lookahead
    
    df['strat_ret'] = df['signal'] * df['size'] * df['ret']
    df['equity'] = (1 + df['strat_ret']).cumprod()
    
    total_ret = df['equity'].iloc[-1] - 1
    annual_ret = (1 + total_ret) ** (252 / len(df)) - 1
    sharpe = df['strat_ret'].mean() / df['strat_ret'].std() * (252 ** 0.5)
    max_dd = (df['equity'] / df['equity'].cummax() - 1).min()
    win_rate = (df[df['signal'] != 0]['strat_ret'] > 0).mean()
    n_trades = (df['signal'] != 0).sum()
    
    metrics = dict(annual=annual_ret, sharpe=sharpe, max_dd=max_dd,
                   win_rate=win_rate, n_days=n_trades)
    
    if not silent:
        for k, v in metrics.items():
            print(f"{k}: {v:.3f}")
    
    return df, metrics
```
